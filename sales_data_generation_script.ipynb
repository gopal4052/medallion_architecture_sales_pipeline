```python
import random
import pandas as pd

names = ["Alice","Bob","Charlie","David","Eva","Frank","Grace","Hank","Ivy","Jack"]
cities = ["Mumbai","mumbai","KARNATAKA","karnataka","Noida","noida",None]
sales_values = [1000,1500,1800,2000,2200,2500,None,"abc"]

data = []

for i in range(10000):
    row = {
        "id": random.randint(1, 5000),
        "name": random.choice(names),
        "age": random.randint(22, 50),
        "city": random.choice(cities),
        "sales": random.choice(sales_values)
    }
    data.append(row)

df = pd.DataFrame(data)

file_path = "/Volumes/data-engineering/raw_data/raw_data/sales_data_large.csv"
df.to_csv(file_path, index=False)

print("Done")
```
